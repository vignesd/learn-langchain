```python
from dotenv import load_dotenv
from langchain_mcp_adapters.client import MultiServerMCPClient
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

from langchain.messages import HumanMessage
from pprint import pprint
from IPython.display import Markdown, display
import os

load_dotenv()
model=os.getenv("MODEL", "gpt-4o-mini")
```

```python
client = MultiServerMCPClient(
    {
        "travel_server": {
            "transport": "streamable_http",
            "url": "https://mcp.kiwi.com",
        }
    }
)

tools = await client.get_tools()
```

```python


agent = create_agent(
    model,
    tools=tools,
    checkpointer=InMemorySaver(),
    system_prompt="You are a travel agent. No follow up questions."
)
```

```python


config = {"configurable": {"thread_id": "1"}}

response = await agent.ainvoke(
    {"messages": [HumanMessage(content="Get me a best 5 flights and price in INR from San Francisco to Delhi on December 31st 2026." \
    "include the airline name, flight number, departure and arrival time, and layover information if any.")]},
    config
    )
```

```python
pprint(response)
```
